```python
import pandas as pd
import numpy as np
```

```python

books = pd.read_csv('processed datasets/Books.csv')
ratings = pd.read_csv('processed datasets/Ratings.csv')

```

```python
merged_data = pd.merge(ratings, books, on='ISBN')
merged_data.head()
```

```python
merged_data = merged_data.iloc[: , : 7 ]
```

##### a book that has been rated 10 highest number of times 

```python
#highest ratins
top_rating = merged_data['Book-Rating'].max()


#book ibn with highest number of 10star ratings
most_rated_books = merged_data[merged_data['Book-Rating'] == top_rating]['ISBN'].value_counts()[:10].index
most_rated_books


top_books = books[books['ISBN'].isin(most_rated_books)]
top_books.iloc[: ,:5]
```

#### customised ratings based suggesttion

```python
ratings = int(input('Enter ratings : '))


#book ibn with highest number of 10star ratings
most_rated_books = merged_data[merged_data['Book-Rating'] == ratings]['ISBN'].value_counts()[:10].index
most_rated_books


top_books = books[books['ISBN'].isin(most_rated_books)]
top_books.iloc[: ,:5]
```
