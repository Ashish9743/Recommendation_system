```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
books = pd.read_csv('processed datasets/books.csv')
ratings = pd.read_csv("processed datasets/ratings.csv")
```

```python
books = books.iloc[: , :5]
books['Year-Of-Publication'] = pd.to_datetime(books['Year-Of-Publication']).dt.year

books.head()
```

```python
ratings = ratings[ratings['Book-Rating'] != 0]
ratings.head()
```

```python
books['combined_features'] = books['Book-Title'] + ' ' + books['Book-Author']
```

```python
tf = TfidfVectorizer(lowercase=True , stop_words='english' , max_features=5000)

tf_matrix = tf.fit_transform(books['combined_features'])

tf_matrix[0].toarray()
```

```python
users_profiles = {}


print('Index of Rated books are : ')
for user in ratings['User-ID'].unique()[:10]:
    
    rated_books_id = ratings[ratings['User-ID']== user]['ISBN'].values
    
    
    for id in rated_books_id:
        if id in books['ISBN'].unique():
            books_ratings = ratings[(ratings['User-ID'] == user ) & (ratings['ISBN'] == id )]['Book-Rating'].values

    rated_books_index = books[books['ISBN'].isin(rated_books_id)].index.tolist()

    print(user , rated_books_index)


    if rated_books_index:
        user_matrix = tf_matrix[rated_books_index]

        weighted_vector = []
        for i ,  rating in enumerate(books_ratings):
           weighted_vector.append(user_matrix[i] * rating)
        
        user_vector = sum(weighted_vector)
        users_profiles[user] = user_vector
    


```

```python
def get_similar_books( user_id =276747  , n_tops = 5):

    users_profile = users_profiles.get(user_id)

    if users_profile is None:
        return None

    already_rated_books = set(ratings[ratings['User-ID'] == user_id]['ISBN'].values)

    users_similarities = cosine_similarity(users_profile, tf_matrix).flatten()
    # print(users_similarities)

    unrated_books = [i for i ,isbn in enumerate(books['ISBN']) if isbn not in already_rated_books ]
    # print(unrated_books)

    unrated_similarities = [(i , users_similarities[i]) for i in unrated_books]
    unrated_similarities.sort(reverse= True)


    top_indeices = [i for i ,sim in unrated_similarities[:n_tops]]
    recomended_books = books.iloc[top_indeices]

    return recomended_books[['ISBN', 'Book-Title', 'Book-Author']]
```
